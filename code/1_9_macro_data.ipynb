```python
import pandas as pd
```

Real Gross Domestic Product (https://fred.stlouisfed.org/series/NGDPRXDCITA)

```python
df_gdp = pd.read_csv('C:/Users/HP/Desktop/Traineeship/data/FRED/real_GDP.csv', index_col=False)

df_gdp['observation_date'] = pd.to_datetime(df_gdp['observation_date'])

df_gdp['year'] = pd.DatetimeIndex(df_gdp['observation_date']).year

df_gdp = df_gdp.loc[(df_gdp['year'] >= 2014) & (df_gdp['year'] <= 2024)].rename(columns = {'NGDPRXDCITA' : 'real_gdp'}).drop(columns = ['observation_date'])

df_gdp = df_gdp.groupby(['year'], as_index = False).agg({'real_gdp' : 'mean'})

df_gdp.to_parquet('datasets/macro/gdp_real.parquet', index=False)
```

Interest rates for house purchases

```python
df_interest = pd.read_csv('C:/Users/HP/Desktop/Traineeship/data/REPORT_07-01-2026 11-22_1767781325081.csv', sep = ';', index_col=False)

df_interest = df_interest.rename(columns = {'Observation date' : 'observation_date', 'Interest rate - loans for house purchases - new business' : 'interest_rate'})

df_interest['observation_date'] = pd.to_datetime(df_interest['observation_date'])

df_interest['year'] = pd.DatetimeIndex(df_interest['observation_date']).year

df_interest = df_interest.loc[(df_interest['year'] >= 2014) & (df_interest['year'] <= 2024)].drop(columns = ['observation_date'])

df_interest = df_interest.groupby(['year'], as_index = False).agg({'interest_rate' : 'mean'})

df_interest.to_parquet('datasets/macro/house_interest.parquet', index=False)
```

Real HPI (https://fred.stlouisfed.org/series/QITR628BIS)

```python
df_hpi = pd.read_csv('C:/Users/HP/Desktop/Traineeship/data/FRED/HPI_FRED.csv', index_col=False)

df_hpi = df_hpi.rename(columns = {'Observation date' : 'observation_date', 'QITR628BIS' : 'hpi'})

df_hpi['observation_date'] = pd.to_datetime(df_hpi['observation_date'])

df_hpi['year'] = pd.DatetimeIndex(df_hpi['observation_date']).year

df_hpi = df_hpi.groupby(['year'], as_index = False).agg({'hpi' : 'mean'})

df_hpi.to_parquet('datasets/macro/hpi.parquet', index=False)
```
